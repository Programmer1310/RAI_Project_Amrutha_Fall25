```python
import pandas as pd
```

```python
file_path = "/content/drive/MyDrive/RAI Project/Project Implementation/Citywide_Payroll_Data_(Fiscal_Year)_20251110.csv"  # replace with your file path
df = pd.read_csv(file_path)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

# Example of column renaming for convenience
df.rename(columns={
    'mid_init': 'middle_name',
    'last_name': 'last_name',
    'first_name': 'first_name'
}, inplace=True)
```

```python
bisg_cols = ['first_name', 'middle_name', 'last_name', 'work_location_borough']
df_bisg = df[bisg_cols].copy()
```

```python
for col in ['first_name', 'middle_name', 'last_name']:
    df_bisg[col] = df_bisg[col].astype(str).str.strip().str.title()

# Replace empty strings or 'nan' with actual NaN
df_bisg.replace({'': pd.NA, 'Nan': pd.NA, 'nan': pd.NA}, inplace=True)

# Optional: drop rows with missing first or last name
#df_bisg = df_bisg.dropna(subset=['first_name', 'last_name'])
```
