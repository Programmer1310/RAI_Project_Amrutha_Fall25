```python
# --- Parameters ---
TARGET_SAMPLE_SIZE = 100000
STRATIFICATION_COLUMN = 'race_code'
# ------------------

# 1. Calculate the proportion of each race_code
# The value_counts(normalize=True) method calculates the percentage/proportion.
race_proportions = combined_df[STRATIFICATION_COLUMN].value_counts(normalize=True)

#

# 2. Calculate the required sample size for each group
# We multiply the TARGET_SAMPLE_SIZE by the proportion and round to the nearest integer.
# The sum might not be exactly 100,000 due to rounding, but it will be very close.
sample_counts = (race_proportions * TARGET_SAMPLE_SIZE).round().astype(int)

# Create an empty list to store the stratified samples
stratified_samples = []

# 3. Iterate through each race group and sample the required number of rows
print("Sampling distribution:")
print(sample_counts)

for race_code, count in sample_counts.items():
    # Filter the original DataFrame for the current race_code
    subset = combined_df[combined_df[STRATIFICATION_COLUMN] == race_code]

    # Sample 'count' rows from this subset
    # 'replace=False' ensures we don't pick the same row twice
    # 'random_state' ensures reproducibility
    sample = subset.sample(n=count, replace=False, random_state=42)

    # Add the sample to our list
    stratified_samples.append(sample)

# 4. Concatenate all the individual race samples into the final DataFrame
df_sampled = pd.concat(stratified_samples).reset_index(drop=True)

# --- Verification ---
# Check the final size
print(f"\nOriginal size: {len(combined_df)}")
print(f"Sampled size: {len(df_sampled)}")

# Verify the final distribution is the same as the original
original_dist = combined_df[STRATIFICATION_COLUMN].value_counts(normalize=True).sort_index()
sampled_dist = df_sampled[STRATIFICATION_COLUMN].value_counts(normalize=True).sort_index()

# Print for comparison (values should be almost identical)
comparison_table = pd.DataFrame({
    'Original Proportion': original_dist,
    'Sampled Proportion': sampled_dist
})
print("\nDistribution Comparison:")
print(comparison_table.mul(100).round(2).astype(str) + '%')

df_sampled.to_csv('stratified_sample_100k.csv', index=False)

print("Successfully saved the new dataset to 'stratified_sample_100k.csv'")
```

```python


def evaluate_cutoff(df, race_col, true_col, cutoff):
    """
    Compute sensitivity and specificity for a given race and cutoff.

    race_col = column with predicted probability for the race (e.g. "prob_black")
    true_col = column with the true race label (e.g. "race_code")
    cutoff   = threshold to test (0.50 ... 1.00)
    """

    # predicted positives (assigned to the race)
    pred_positive = df[race_col] > cutoff

    # true positives (actual race members)
    true_positive = df[true_col] == race_col.replace("prob_", "")

    # sensitivity = TP / actual positives
    sensitivity = (pred_positive & true_positive).sum() / true_positive.sum()

    # specificity = TN / actual negatives
    specificity = ((~pred_positive) & (~true_positive)).sum() / (~true_positive).sum()

    return sensitivity, specificity
```

```python
def find_optimal_cutoff(df, race_col, true_col):
    cutoffs = np.arange(0.50, 1.01, 0.01)

    results = []

    for cutoff in cutoffs:
        sens, spec = evaluate_cutoff(df, race_col, true_col, cutoff)
        distance = np.sqrt((1 - sens)**2 + (1 - spec)**2)

        results.append({
            "cutoff": cutoff,
            "sensitivity": sens,
            "specificity": spec,
            "distance": distance
        })

    results_df = pd.DataFrame(results)

    # smallest distance = optimal cutoff
    best_row = results_df.loc[results_df["distance"].idxmin()]

    return best_row, results_df
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score

# --------------------------
# 1. Load your BISG CSV
# --------------------------
file_path = "/content/drive/MyDrive/RAI Project/Project Implementation/bisg_done_sample.csv"  # replace with your file path
df = pd.read_csv(file_path)

# Ensure probability columns are numeric, coercing errors to NaN
prob_cols_to_check = ["prob_white", "prob_black", "prob_hisp", "prob_asian"]
for col in prob_cols_to_check:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# --------------------------
# 2. Functions for optimal cutoff
# --------------------------
def evaluate_cutoff(df, prob_col_name, true_race_code_label, cutoff):
    pred_positive = df[prob_col_name] > cutoff
    true_positive = df['race_code'] == true_race_code_label

    total_true_positives = true_positive.sum()
    if total_true_positives == 0:
        sensitivity = np.nan
    else:
        sensitivity = (pred_positive & true_positive).sum() / total_true_positives

    total_true_negatives = (~true_positive).sum()
    if total_true_negatives == 0:
        specificity = np.nan
    else:
        specificity = ((~pred_positive) & (~true_positive)).sum() / total_true_negatives

    return sensitivity, specificity

def find_optimal_cutoff(df, prob_col_name, true_race_code_label):
    cutoffs = np.arange(0.50, 1.01, 0.01)
    results = []

    for cutoff in cutoffs:
        sens, spec = evaluate_cutoff(df, prob_col_name, true_race_code_label, cutoff)
        if not np.isnan(sens) and not np.isnan(spec):
            distance = np.sqrt((1 - sens)**2 + (1 - spec)**2)
        else:
            distance = np.nan
        results.append({
            "cutoff": cutoff,
            "sensitivity": sens,
            "specificity": spec,
            "distance": distance
        })

    results_df = pd.DataFrame(results)
    if results_df["distance"].isnull().all():
        return pd.Series({"cutoff": np.nan}), results_df
    else:
        best_row = results_df.loc[results_df["distance"].idxmin()]
        return best_row, results_df

# --------------------------
# 3. Compute optimal cutoffs for each race
# --------------------------
race_prob_cols_mapping = {
    "prob_white": "W",
    "prob_black": "B",
    "prob_hisp": "HL",
    "prob_asian": "A"
}

optimal_cutoffs = {}
for prob_col, race_code in race_prob_cols_mapping.items():
    best_row, _ = find_optimal_cutoff(df, prob_col, race_code)
    optimal_cutoffs[prob_col] = best_row["cutoff"]
    print(f"{prob_col} optimal cutoff: {best_row['cutoff']:.2f}")

# --------------------------
# 4. Thresholding predictions
# --------------------------
def predict_race_threshold(row, race_prob_cols_mapping, optimal_cutoffs):
    pred_race = 'Unknown'
    for prob_col, race_code in race_prob_cols_mapping.items():
        cutoff = optimal_cutoffs[prob_col]
        if pd.notna(row[prob_col]) and row[prob_col] >= cutoff:
            pred_race = race_code
            break  # assign first race that passes threshold
    return pred_race
    # Force all probability columns to numeric
    for col in ['prob_white', 'prob_black', 'prob_hisp', 'prob_asian']:
      df[col] = pd.to_numeric(df[col], errors='coerce')

df['pred_race_threshold'] = df.apply(lambda row: predict_race_threshold(row, race_prob_cols_mapping, optimal_cutoffs), axis=1)

# --------------------------
# 5. Weighted predictions
# --------------------------
'''Monte-Carlo assignment of weights for Weighing:
Take the probability distribution of a row
Example:
white 0.40, black 0.10, hisp 0.45, asian 0.05

Treat these probabilities like weights.

Randomly “draw” one race according to these weights.
E.g., there is:
- 40% chance the draw is White
- 10% chance Black
- 45% chance Hispanic
- 5% chance Asian

Assign the race that was drawn.Repeat this for all 100,000 rows.'''

def predict_race_weighted(row, race_prob_cols_mapping):
    races = list(race_prob_cols_mapping.values())
    prob_cols = list(race_prob_cols_mapping.keys())

    #probs = row[prob_cols].values
    probs = row[prob_cols].astype(float).values


    # Replace NaN values with 0.0 before normalization
    probs_cleaned = np.nan_to_num(probs, nan=0.0)

    total_prob_sum = probs_cleaned.sum()

    # If the sum of probabilities is zero (or very close), return 'Unknown'
    if total_prob_sum <= 1e-9:
        return 'Unknown'

    normalized_probs = probs_cleaned / total_prob_sum

    return np.random.choice(races, p=normalized_probs)

df['pred_race_weighted'] = df.apply(lambda row: predict_race_weighted(row, race_prob_cols_mapping), axis=1)

# --------------------------
# 6. Compare accuracy and coverage
# --------------------------
# Thresholding
mask = df['pred_race_threshold'] != 'Unknown'
accuracy_threshold = accuracy_score(df.loc[mask, 'race_code'], df.loc[mask, 'pred_race_threshold'])
coverage_threshold = mask.mean()  # fraction of rows classified

# Weighted
accuracy_weighted = accuracy_score(df['race_code'], df['pred_race_weighted'])

print("\n--- Comparison ---")
print(f"Thresholding Accuracy: {accuracy_threshold:.3f}")
print(f"Thresholding Coverage: {coverage_threshold:.3f}")
print(f"Weighted Accuracy: {accuracy_weighted:.3f}")
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np
import matplotlib.pyplot as plt

# Correct labels
race_labels = ["W", "B", "HL", "A"]

def plot_normalized_cm(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred, labels=race_labels)
    cm_normalized = cm.astype(float) / cm.sum(axis=1, keepdims=True)

    plt.figure(figsize=(6,6))
    plt.imshow(cm_normalized, aspect='auto')
    plt.title(title)
    plt.colorbar()

    plt.xticks(np.arange(len(race_labels)), race_labels, rotation=45)
    plt.yticks(np.arange(len(race_labels)), race_labels)

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    # Annotate values
    for i in range(len(race_labels)):
        for j in range(len(race_labels)):
            val = cm_normalized[i, j]
            plt.text(j, i, f"{val:.2f}", ha="center", va="center")

    plt.show()


# Thresholding
plot_normalized_cm(df["race_code"], df["pred_race_threshold"], "Normalized Confusion Matrix — Thresholding")

# Weighted
plot_normalized_cm(df["race_code"], df["pred_race_weighted"], "Normalized Confusion Matrix — Weighted Sampling")
```
