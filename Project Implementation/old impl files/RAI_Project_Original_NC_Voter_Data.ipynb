```python
# --- Parameters ---
TARGET_SAMPLE_SIZE = 100000
STRATIFICATION_COLUMN = 'race_code'
# ------------------

# 1. Calculate the proportion of each race_code
# The value_counts(normalize=True) method calculates the percentage/proportion.
race_proportions = combined_df[STRATIFICATION_COLUMN].value_counts(normalize=True)

#

# 2. Calculate the required sample size for each group
# We multiply the TARGET_SAMPLE_SIZE by the proportion and round to the nearest integer.
# The sum might not be exactly 100,000 due to rounding, but it will be very close.
sample_counts = (race_proportions * TARGET_SAMPLE_SIZE).round().astype(int)

# Create an empty list to store the stratified samples
stratified_samples = []

# 3. Iterate through each race group and sample the required number of rows
print("Sampling distribution:")
print(sample_counts)

for race_code, count in sample_counts.items():
    # Filter the original DataFrame for the current race_code
    subset = combined_df[combined_df[STRATIFICATION_COLUMN] == race_code]

    # Sample 'count' rows from this subset
    # 'replace=False' ensures we don't pick the same row twice
    # 'random_state' ensures reproducibility
    sample = subset.sample(n=count, replace=False, random_state=42)

    # Add the sample to our list
    stratified_samples.append(sample)

# 4. Concatenate all the individual race samples into the final DataFrame
df_sampled = pd.concat(stratified_samples).reset_index(drop=True)

# --- Verification ---
# Check the final size
print(f"\nOriginal size: {len(combined_df)}")
print(f"Sampled size: {len(df_sampled)}")

# Verify the final distribution is the same as the original
original_dist = combined_df[STRATIFICATION_COLUMN].value_counts(normalize=True).sort_index()
sampled_dist = df_sampled[STRATIFICATION_COLUMN].value_counts(normalize=True).sort_index()

# Print for comparison (values should be almost identical)
comparison_table = pd.DataFrame({
    'Original Proportion': original_dist,
    'Sampled Proportion': sampled_dist
})
print("\nDistribution Comparison:")
print(comparison_table.mul(100).round(2).astype(str) + '%')

df_sampled.to_csv('stratified_sample_100k.csv', index=False)

print("Successfully saved the new dataset to 'stratified_sample_100k.csv'")
```

```python


def evaluate_cutoff(df, race_col, true_col, cutoff):
    """
    Compute sensitivity and specificity for a given race and cutoff.

    race_col = column with predicted probability for the race (e.g. "prob_black")
    true_col = column with the true race label (e.g. "race_code")
    cutoff   = threshold to test (0.50 ... 1.00)
    """

    # predicted positives (assigned to the race)
    pred_positive = df[race_col] > cutoff

    # true positives (actual race members)
    true_positive = df[true_col] == race_col.replace("prob_", "")

    # sensitivity = TP / actual positives
    sensitivity = (pred_positive & true_positive).sum() / true_positive.sum()

    # specificity = TN / actual negatives
    specificity = ((~pred_positive) & (~true_positive)).sum() / (~true_positive).sum()

    return sensitivity, specificity
```

```python
def find_optimal_cutoff(df, race_col, true_col):
    cutoffs = np.arange(0.50, 1.01, 0.01)

    results = []

    for cutoff in cutoffs:
        sens, spec = evaluate_cutoff(df, race_col, true_col, cutoff)
        distance = np.sqrt((1 - sens)**2 + (1 - spec)**2)

        results.append({
            "cutoff": cutoff,
            "sensitivity": sens,
            "specificity": spec,
            "distance": distance
        })

    results_df = pd.DataFrame(results)

    # smallest distance = optimal cutoff
    best_row = results_df.loc[results_df["distance"].idxmin()]

    return best_row, results_df
```

```python
import numpy as np
import pandas as pd

# Redefining evaluate_cutoff within this cell to fix the error
def evaluate_cutoff(df, prob_col_name, true_race_code_label, cutoff):
    """
    Compute sensitivity and specificity for a given race and cutoff.

    prob_col_name = column with predicted probability for the race (e.g. "prob_black")
    true_race_code_label = the actual race code label in the 'race_code' column (e.g. "B" for Black)
    cutoff   = threshold to test (0.50 ... 1.00)
    """
    pred_positive = df[prob_col_name] > cutoff

    true_positive = df['race_code'] == true_race_code_label


    total_true_positives = true_positive.sum()
    if total_true_positives == 0:
        sensitivity = np.nan
    else:
        sensitivity = (pred_positive & true_positive).sum() / total_true_positives

    total_true_negatives = (~true_positive).sum()
    if total_true_negatives == 0:
        specificity = np.nan
    else:
        specificity = ((~pred_positive) & (~true_positive)).sum() / total_true_negatives

    return sensitivity, specificity

# Redefining find_optimal_cutoff within this cell to fix the error
def find_optimal_cutoff(df, prob_col_name, true_race_code_label):
    cutoffs = np.arange(0.50, 1.01, 0.01)
    results = []

    for cutoff in cutoffs:
        sens, spec = evaluate_cutoff(df, prob_col_name, true_race_code_label, cutoff)
        if not np.isnan(sens) and not np.isnan(spec):
            distance = np.sqrt((1 - sens)**2 + (1 - spec)**2)
        else:
            distance = np.nan

        results.append({
            "cutoff": cutoff,
            "sensitivity": sens,
            "specificity": spec,
            "distance": distance
        })

    results_df = pd.DataFrame(results)

    if results_df["distance"].isnull().all():
        print(f"Warning: All distances are NaN for {prob_col_name}. Cannot find optimal cutoff.")
        # Return a Series with NaN values for consistency
        return pd.Series({"cutoff": np.nan, "sensitivity": np.nan, "specificity": np.nan, "distance": np.nan}), results_df
    else:
        best_row = results_df.loc[results_df["distance"].idxmin()]
        return best_row, results_df


# Mapping for probability columns to actual race codes in df_with_zcta['race_code']
race_prob_cols_mapping = {
    "prob_white": "W",
    "prob_black": "B",
    "prob_hisp": "HL", # Assuming 'HL' for Hispanic based on previous cells
    "prob_asian": "A"  # Assuming 'A' for Asian based on typical race codes
}

optimal_cutoffs = {}
print("size:" + str(df_with_zcta.size))

for prob_col_name, true_race_code_label in race_prob_cols_mapping.items():
    print(f"\nEvaluating {prob_col_name} against actual race code '{true_race_code_label}'...")
    best, roc_table = find_optimal_cutoff(df_with_zcta, prob_col_name, true_race_code_label)
    optimal_cutoffs[prob_col_name] = best["cutoff"]
    print(best)
```
